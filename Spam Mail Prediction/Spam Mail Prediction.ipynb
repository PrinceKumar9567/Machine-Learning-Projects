Importing The Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import cross_val_score
```

Data Loading and Preprocessing

```python
#Loading the Data
data = pd.read_csv(url)
```

```python
#Printing first Five rows of the data
data.head()
```

```python
data.shape
```

```python
#Checking for any missing values
data.isnull().sum()
```

```python
#Encoding the Category column
le = LabelEncoder()
data['Numeric_Category'] = le.fit_transform(data['Category'])
```

```python
data.head()
```

Seprating The data to the Features and Target Column

```python
X = data['Message']
Y = data['Numeric_Category']
```

Spam Mail--1

Ham Mail--0

Feature Extraction

```python
#Now converting the texual data column Message to Numerical data
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
vectorizer.fit(X)
X = vectorizer.transform(X)
```

```python
print(X)
```

```python
#Loading the Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
models = [LogisticRegression(max_iter = 10000),DecisionTreeClassifier(max_depth = 5),SVC(kernel = 'linear',C = 1.0),KNeighborsClassifier(n_neighbors = 5,weights = 'uniform')]
```

```python
def compare_model_cross_validation():
    for model in models:
        cv_score = cross_val_score(model,X,Y,cv = 5)
        mean_Accuracy = round((sum(cv_score)/len(cv_score))*100,2)
        print('Cross Validation Score for the ',model,'=',cv_score)
        print('Mean Accuracy of the ',model,'=',mean_Accuracy)
        print('---------------------------------------------------------------------------------')
        
        
```

```python
compare_model_cross_validation()
```

So as we can see that in betwen all of these four models Support Vector Classifier got the Highest Accuracy of 98.17%
